# Discount vs Profit (Scatter Plot)

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x='Discount',
    y='Profit',
    )
plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.show()
```

### Insight
1. Higher Discounts Correlate with Lower/Negative Profits: As the discount rate increases beyond 0.4–0.5, profit values trend downward and frequently dip into negative territory, suggesting aggressive discounting erodes margins significantly.
2. Zero or Low Discounts Yield the Highest Profits: The highest profit values (6,000–8,000) are concentrated at a discount of 0.0, indicating that full-price sales are the most profitable transactions.
3. Losses Begin Around 0.4–0.5 Discount: Negative profits start appearing consistently at discount levels of 0.4 and above, with losses reaching as low as –6,000 at higher discount bands (0.7–0.8).
4. Wide Profit Variance at Low Discounts: At discounts between 0.0 and 0.2, profit values are highly spread out (ranging from near 0 to 8,000+), suggesting other factors like product category or order size also influence profitability.
5. High Discounts Cluster Near Zero or Below: At discounts of 0.6, 0.7, and 0.8, most data points are near or below zero profit, with very few positive outcomes — making these discount levels financially unsustainable.
Overall Takeaway: Discounts above 40% are a strong predictor of unprofitable transactions and should be reviewed or capped in business strategy.


# Correlation Heatmap

```python
plt.figure(figsize=(8,5))

corr = df[['Sales', 'Quantity', 'Discount', 'Profit']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)

plt.title('Correlation Heatmap')
plt.show()
```

### Insight
1. Sales and Profit have the Strongest Positive Correlation (0.48): This is the most notable relationship — higher sales revenue is moderately associated with higher profit, which is expected but not guaranteed due to cost factors.
2. Sales and Quantity are Weakly Positively Correlated (0.2): A slight positive relationship exists, meaning selling more units tends to marginally increase sales revenue, but the effect is not strong.
3. Discount negatively impacts Profit (–0.22): Discount and Profit share a negative correlation, confirming (alongside the scatter plot) that giving higher discounts reduces profitability. This is a critical business finding.
4. Discount and Sales are Nearly Uncorrelated (–0.028): Discounts don't meaningfully drive more sales revenue, suggesting discounting is not effectively boosting top-line performance either.
5. Quantity and Profit are Barely Correlated (0.066): Selling more units does not reliably translate to more profit, likely due to low-margin or discounted bulk orders.
6. Quantity and Discount are Almost Independent (0.0086): Discount level has virtually no relationship with the quantity of items ordered.

Overall Takeaway: The data strongly suggests that discounts hurt profit without meaningfully boosting sales or quantity, making discount strategy the most actionable area for business improvement.

# Sales Distribution

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Sales'], bins=30, kde=True)

plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Frequency')
plt.show()
```

### Insight
1. Heavily Right-Skewed Distribution: The vast majority of transactions are concentrated at very low sales values (near 0–1,000), while a long tail stretches toward 20,000. This is a classic right-skewed (positively skewed) distribution.
2. Most Transactions are Small-Value Sales: The peak frequency exceeds 17,500 at the lowest sales bin, meaning the overwhelming majority of orders involve relatively small purchase amounts.
3. High-Value Sales are Rare but Exist: Sales beyond 5,000 are very infrequent, and transactions above 10,000–20,000 are extreme outliers — likely large bulk or enterprise orders.
4. Rapid Frequency Drop-Off: Frequency drops sharply after the first bin (~0–500 range), then tapers off very gradually, indicating that mid-to-high value sales are consistently uncommon across the dataset.
5. Presence of High-Value Outliers: The extended tail up to ~20,000 suggests a small number of very high-value transactions that could disproportionately influence total revenue despite being rare.

Overall Takeaway: The business is driven largely by high-volume, low-value transactions. A small number of high-value orders exist and likely contribute significantly to total revenue — identifying and nurturing those high-value customers could be a key growth strategy.


# Profit Distribution (Box Plot)

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=df['Profit'])

plt.title('Profit Distribution')
plt.xlabel('Profit')
plt.show()
```

### Insight
1. Profits are Heavily Concentrated Near Zero: The densest cluster of data points lies between approximately –500 and +2000, indicating that most transactions generate modest profits or near-breakeven results.
2. Distribution is Slightly Right-Skewed: While most data points cluster just above zero, there is a longer tail extending toward +8,000, meaning a few transactions generate very high profits.
3. Significant Losses Exist (Left Tail): The distribution extends as far left as –6,000, confirming that some transactions result in substantial losses — consistent with the heavy discounting patterns seen in the previous graphs.
4. Outliers on Both Ends: Isolated data points appear at both extremes (around –6,000 and +8,000), suggesting occasional exceptional transactions — either very profitable deals or severely loss-making ones.
5. Majority of Transactions are Profitable: The bulk of data points sit on the positive side of zero, which is a healthy sign — losses exist but are not the dominant outcome across all transactions.
6. Thin Spread Beyond ±3000: Very few transactions fall outside the –3,000 to +3,000 range, meaning extreme profit or loss events are rare and largely driven by outlier orders.

Overall Takeaway: While the business is generally profitable at the transaction level, the presence of deep losses (down to –6,000) highlights that certain orders — likely heavily discounted ones — are significantly dragging down overall profitability.
